<h4>Q1. A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a
subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using
broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

```python
import numpy as np
marks = np.array([
    [92, 99, 95],
    [78, 85, 94],
    [88, 90, 89],
    [96, 95, 93]
])
bonus = np.array([5, 2, 8])
updated_marks = marks + bonus
capped_count = np.sum(updated_marks > 100)
final_marks = np.clip(updated_marks, a_min=None, a_max=100)

print(final_marks)
print(capped_count)
```

<h4>Q9. Award medals to the scores below using nested np.where: 'Gold' for 90 and above, 'Silver' for 75
and above, 'Bronze' for 60 and above, and 'None' otherwise. Then count how many students received
each medal.
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

```python
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])
medals = np.where(scores >= 90, 'Gold',
         np.where(scores >= 75, 'Silver',
         np.where(scores >= 60, 'Bronze', 'None')))
unique_medals, counts = np.unique(medals, return_counts=True)
medal_counts = dict(zip(unique_medals, counts))
print( scores)
print( medals)
for medal, count in medal_counts.items():
    print(medal,count)
```

<h4>Q14. Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column
vectors, join them into a single two-column table, attach a third column holding each row's total, then
multiply the three-column table by w = [0.5, 0.2, 0.3] to obtain one score per row. Finally use
np.where to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of
every intermediate array.
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

```python
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
w = np.array([0.5, 0.2, 0.3])
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)
print(hours_col.shape, attendance_col.shape)

table_2col = np.hstack((hours_col, attendance_col))
print( table_2col.shape)

row_totals = np.sum(table_2col, axis=1, keepdims=True)
print( row_totals.shape)

table_3col = np.hstack((table_2col, row_totals))
print(table_3col.shape)

scores = table_3col @ w
print( scores.shape)

avg_score = np.mean(scores)
binary_labels = np.where(scores > avg_score, 1, 0)
print( binary_labels.shape)


print( scores)
print( avg_score)
print( binary_labels)
```

<h4>Q19. Given the marks matrix below (5 students x 4 subjects), first replace every mark below 35 with 35,
then compute each student's total, and finally print the index of the topper along with their total.
marks = np.array([[45, 28, 67, 80],
 [33, 90, 55, 41],
 [72, 66, 30, 58],
 [88, 79, 91, 84],
 [25, 40, 38, 52]])

```python
marks = np.array([[45, 28, 67, 80],[33, 90, 55, 41],[72, 66, 30, 58],[88, 79, 91, 84],[25, 40, 38, 52]])

updated_marks = np.clip(marks, a_min=35, a_max=None)

totals = updated_marks.sum(axis=1)

topper_index = totals.argmax()
topper_total = totals[topper_index]

print("Updated Marks:\n", updated_marks)
print("\nStudent Totals:", totals)
print("Topper Index:", topper_index)
print("Topper Total:", topper_total)
```
