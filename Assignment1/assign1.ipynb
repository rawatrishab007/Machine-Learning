<h1>Question 1 · The Strict Username Validator</h1>
<p>Write a function that takes a username string as input and validates it. To be valid, the username must be at
least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and
numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase
version of the username. If it is invalid, return the string "Invalid".

```python
string='CodeMaster99'
clean=string.strip()
if len(clean)>5 and clean.isalnum():
    print(clean.lower())
else:
    print('Invalid')
```

<h1>Question 2 · The Spy Code Reverser</h1>
<p>You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the
order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

```python
secret_code = "charlie-bravo-alpha" 
rs=secret_code.split('-')
res=rs[::-1]
print(" ".join(res).upper())
```

<h1>Question 3 · The Messy Database Cleaner</h1>
<p>You have received a raw string of names from a database exported as a semicolon-separated list. The data is
messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every
name is properly Title Cased, extra spaces are removed, and the names are joined into a single string
separated by a comma and a space.

```python
raw_names = " john doe ; jAnE sMiTh; ALAN tUrInG "
clean=raw_names.strip()
rs=clean.split(';')
res=",".join(rs).title()
print(res)
```

<h1>Question 5 · The Palindrome Sentence Checker</h1>
<p>Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces
and non-alphanumeric characters, convert everything to lowercase, and check whether it reads the same
forwards and backwards.</p>

```python
sentence = "Was it a car or a cat I saw?"
clean=''
for ch in sentence:
    if ch.isalnum():
        clean+=ch.lower()
ans=clean[::-1]
if ans==clean:
    print("true")
else:
    print("false")
```

<h1>Question 6 · The Hashtag Extractor</h1>
<p>Given a string representing a social media post, extract all the hashtags. The extracted hashtags must be
completely lowercase, and the # symbol should be removed. Store them in a list. Ignore any words that are not
hashtags</p>


```python
post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"
pst=post.strip().split()
ans=[]
for ch in pst:
    if ch.startswith('#'):
        ans.append(ch[1:].lower())
print(ans)
```

<h1>Question 7 · The File Extension Finder</h1>
<p>You are given a single comma-separated string containing various filenames. Extract just the file extensions
(without the dot), ensure there are no duplicates, and return them as a list of strings strictly in UPPERCASE.</p>

```python
filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
res=[]
for ch in filenames.split(','):
    ext=ch.strip().split('.')[-1].upper()
    res.append(ext)
print(res)
```

<h1>Question 9 · The Anagram Detective</h1>
<p>Write a function to check if two given strings are anagrams of each other — they contain the exact same letters
in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as
the same.

```python
str1="Dormitory"
str2="Dirty Room"
clean1 = str1.replace(" ", "").lower()
clean2 = str2.replace(" ", "").lower()
if sorted(clean1)==sorted(clean2):
    print('true')
else:
    print('false')
```

<h1>Question 10 · The Messy Phone Number Formatte</h1>
<p>You are building a CRM system and user-submitted phone numbers are a mess — they contain spaces,
parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by
removing all non-digit characters. Once cleaned, validate the number:
• If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".
• If it contains anything other than 10 digits, return "Invalid Number".
Note: Case 1 fails because after removing symbols it has 11 digits, due to the leading “1”.

```python
raw_num = "555-987-6543" 
digits=""
for ch in raw_num:
    if ch.isdigit():
        digits+=ch
if len(digits)!=10:
    print("Invlaid")
part1 = digits[0:3]   #pehle k 3 digit 
part2 = digits[3:6]   # Beech ke 3 digits
part3 = digits[6:10]    # last k digits
print('('+part1+')'+part2+"-"+part3)
```

<h1>Question 11 · The Product SKU Inspector</h1>
<p>An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly
3 uppercase letters, a single hyphen, and exactly 4 digits (e.g. "ABC-1234").
Write a function that receives a messy SKU string. First clean it by stripping leading/trailing spaces and
converting it to uppercase. Then validate it using string methods. Return "Valid: [Cleaned_SKU]" if it
passes, or "Invalid SKU" if it fails any formatting rule.
Note: Use isalpha(), isdigit() etc. — no re module. Case 2 fails (2 letters, 5 digits); Case 3 fails (special character).

```python

sku = " xYz-9876 " 
clean_sku = sku.strip().upper()
if len(clean_sku) != 8:
    print( "Invalid SKU")
parts = clean_sku.split("-")
if len(parts) != 2:
    print( "Invalid SKU")
letters = parts[0]
digits = parts[1]
if len(letters) == 3 and letters.isalpha() and len(digits) == 4 and digits.isdigit():
    print("Valid: " + clean_sku)
else:
    print ("Invalid SKU")
```
