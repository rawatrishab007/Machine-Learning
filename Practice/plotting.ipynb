```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid", palette="dark", font_scale=1.05)
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
temp=pd.read_csv('mpg.csv')
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=df,x='Fare')
plt.title('Fare distribution')

plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=df,x='Age',bins=100)
plt.title("Age wala")
plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=temp,x='mpg',bins=25)
plt.title("mpg data set")
print(temp['mpg'].agg(['mean','median']))
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data=temp,x='origin')
plt.show()
print(temp['origin'].value_counts())
```

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='Survived',y='Age',hue='Sex')
plt.title('Fare by survival')
plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.scatterplot(data=df,x='Age',y='Fare',hue='Survived')
plt.title("Scatterplot")
plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.countplot(data=df,x='Pclass',hue='Survived')
plt.show()
```

```python
sns.pairplot(df[['Age', 'Fare', 'Pclass', 'Survived']].dropna(), hue='Survived', diag_kind='hist')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.heatmap(df.drop(columns='PassengerId').corr(numeric_only=True),annot=True)
plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.heatmap(temp.corr(numeric_only=True),annot=True,fmt='.2f',cmap='RdBu_r')
plt.show()
```
