```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid", palette="dark", font_scale=1.05)
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
temp=pd.read_csv('mpg.csv')
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=df,x='Fare')
plt.title('Fare distribution')

plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=df,x='Age',bins=100)
plt.title("Age wala")
plt.show()
```

```python
plt.Figure(figsize=(7,4))
sns.histplot(data=temp,x='mpg',bins=25)
plt.title("mpg data set")
print(temp['mpg'].agg(['mean','median']))
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data=temp,x='origin')
plt.show()
print(temp['origin'].value_counts())
```
