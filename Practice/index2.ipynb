```python
import pandas as pd
```

```python
data = {
    'Name': ['Ravi', 'Aman', 'Neha', 'Priya', 'Karan', 'Simran', 'Arjun', 'Meera'],
    'Marks': [85, 72, None, 91, 68, None, 78, 88],
    'Age': [20, 21, 19, None, 20, 22, None, 21],
    'City': ['Delhi', 'Pune', 'Mumbai', 'Delhi', None, 'Pune', 'Jaipur', 'Mumbai']
}
df = pd.DataFrame(data)
print(df)
```

```python
df.isnull()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.dropna(subset="City")
```

```python
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
print(df)
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
print(df)
```

```python
df['City']=df['City'].fillna('Unknown')
print(df)
```
