<h4>Q1. A user types their city carelessly: " nEw dELhi ". Clean it and print exactly New Delhi. Also
validate that the city contains only alphabets and spaces — print Invalid city otherwise.

```python
city = "   nEw   dELhi   "
words = city.strip().split()
clean_city = " ".join(words).title()
is_valid = True
for ch in clean_city:
    if not (ch.isalpha() or ch.isspace()):
        is_valid = False
        break
if is_valid and len(clean_city) > 0:
    print(clean_city)
else:
    print("Invalid city")
```

<h4>Q2. Prices arrive as one messy string: " 120, 45 , 300 ,80, 95 ". Convert it into a list of integers, then
print the total bill, the highest price, and the average price with 2 decimal places (use an f-string with :.2f).

```python
prices_str = " 120, 45 , 300 ,80, 95 "
prices = [int(p.strip()) for p in prices_str.split(",")]
total = sum(prices)
highest = max(prices)
average = total / len(prices)
print(total)
print(highest)
print(average)
```

<h4>Q3. Validate a username entered by the user (e.g. " Rahul_123 "). Rules:
• clean the spaces and convert to lowercase
• must be alphanumeric (letters and digits only — no symbols)
• must start with an alphabet
• length must be between 5 and 12 characters
Print Valid username or Invalid username. ("Rahul_123" is invalid because of the underscore;
"rahul123" is valid.)

```python
raw_username = " Rahul_123 "
username = raw_username.strip().lower()
is_valid = True
if not (5 <= len(username) <= 12):
    is_valid = False
elif not username.isalnum():
    is_valid = False
elif not username[0].isalpha():
    is_valid = False
if is_valid:
    print("Valid username")
else:
    print("Invalid username")
```

<h4>Q4. For a given sentence, print: (a) the total number of words, (b) how many words end with a vowel, and (c)
the longest word in the sentence.

```python
sentence = "A quick brown fox jumped over the lazy sleeping dog"
words = sentence.split()
total_words = len(words)
vowels = "aeiouAEIOU"
vowel_end_count = 0

for word in words:
    clean_word = word.strip(".,!?;:")
    if clean_word and clean_word[-1] in vowels:
        vowel_end_count += 1
longest_word = max(words, key=len).strip(".,!?;:")
print(f"Total words: {total_words}")
print(f"Words ending with a vowel: {vowel_end_count}")
print(f"Longest word: {longest_word}")
```

<h4>Count the frequency of each character in a string, ignoring spaces and treating uppercase and lowercase
as the same letter. Example: "Aba c" → {'a': 2, 'b': 1, 'c': 1}.

```python
text = "rishab"
clean_text = text.replace(" ", "").lower()
freq = {}
for ch in clean_text:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
```

<h4>Invert the dictionary below so that values become keys. If two days share the same value, store both days
in a list:

```python
sales = {"mon": 2000, "tue": 3050, "wed": 1200}
inverted = {}
for day, amount in sales.items():
    if amount not in inverted:
        inverted[amount] = [day]
    else:
        inverted[amount].append(day)
print(inverted)
```

<h4>Marks of two tests are stored in two dictionaries. Merge them into one dictionary — if a student appears in
both, sum the marks:

```python
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

merged = test1.copy()

for name, marks in test2.items():
    merged[name] = merged.get(name, 0) + marks

print(merged)
```

<h4>. Sort the employees below by salary (highest first) using sorted() with key=lambda, and print each line
using an f-string in the format shown:

```python
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]
sorted_employees = sorted(employees, key=lambda emp: emp[1], reverse=True)
for name, salary in sorted_employees:
    print(f"{name} earns Rs. {salary}")
```
